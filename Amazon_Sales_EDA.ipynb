```python
df['Year'] = df['OrderDate'].dt.year
df['Month'] = df['OrderDate'].dt.month_name()
```

```python
monthly_sales = df.groupby(df['OrderDate'].dt.month_name())['TotalAmount'].sum()
monthly_sales.plot(kind='line', marker='o', figsize=(8,5))

plt.title('Monthly Sales Trend')
plt.xlabel('Total Sales')
plt.ylabel('Month')
plt.grid(True)


plt.tight_layout()
plt.show()
print(monthly_sales)
```

Business Insight:
Sales increased during the middle of the year.
 Some months recorded noticeably lower sales, suggesting seasonal demand.
 Marketing campaigns can be planned around high-performing months.

```python
yearly_sales = df.groupby(df['OrderDate'].dt.year)['TotalAmount'].sum()
yearly_sales.plot(kind='line', marker='o', figsize=(8,5))

plt.title('Yearly Sales Trend')
plt.xlabel('Year')
plt.ylabel('Total Sales')

plt.grid(True)
plt.show()
print(yearly_sales)
```

Business Insight:
The yearly sales trend indicates significant fluctuations over time. Sales peaked in 2020, declined in 2021, recovered in 2023, and dropped again in 2024. This suggests that sales performance was inconsistent across the years.
